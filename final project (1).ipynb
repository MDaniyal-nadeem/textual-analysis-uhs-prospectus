```python
from collections import Counter
import math
import matplotlib.pyplot as plt

TOTAL_PAGES = 83  # confirmed number of pages

# Load final corrected text
with open("uhs_text_final.txt", "r", encoding="utf-8") as f:
    words = f.read().lower().split()

# Split words equally into pages
words_per_page = math.ceil(len(words) / TOTAL_PAGES)
pages = [
    words[i * words_per_page : (i + 1) * words_per_page]
    for i in range(TOTAL_PAGES)
]

# Page-wise top words & plots
for page_no, page_words in enumerate(pages, start=1):
    counter = Counter(page_words)
    
    # Get top 20 most frequent words
    top_words = counter.most_common(20)
    words_list = [w for w, c in top_words]
    counts_list = [c for w, c in top_words]

    # Print top words
    print("\n==============================")
    print(f"PAGE {page_no} TOP 20 WORDS")
    print("==============================")
    for word, freq in top_words:
        print(f"{word}: {freq}")

    # Plot bar graph
    plt.figure(figsize=(10,5))
    plt.bar(words_list, counts_list, color='skyblue')
    plt.title(f"Page {page_no} - Top 20 Words")
    plt.xlabel("Words")
    plt.ylabel("Frequency")
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()
```

```python
#######TASK2 DESCRIPTIVE STATISTICS#######
import numpy as np
import math
from collections import Counter
import matplotlib.pyplot as plt

TOTAL_PAGES = 83  # Confirmed number of pages

# Load final corrected text
with open("uhs_text_final.txt", "r", encoding="utf-8") as f:
    all_words = f.read().lower().split()

# Split words equally into pages
words_per_page = math.ceil(len(all_words) / TOTAL_PAGES)
pages = [
    all_words[i * words_per_page : (i + 1) * words_per_page]
    for i in range(TOTAL_PAGES)
]

# Page-wise descriptive statistics (without removing stopwords)
for page_no, page_words in enumerate(pages, start=1):
    if len(page_words) == 0:
        continue  # skip empty pages

    counter = Counter(page_words)
    frequencies = list(counter.values())

    total_words = sum(frequencies)
    unique_words = len(counter)
    mean_freq = np.mean(frequencies)
    median_freq = np.median(frequencies)
    std_freq = np.std(frequencies)
    min_freq = np.min(frequencies)
    max_freq = np.max(frequencies)

    print("\n==============================")
    print(f"PAGE {page_no} DESCRIPTIVE STATISTICS (ALL WORDS)")
    print("==============================")
    print(f"Total words: {total_words}")
    print(f"Unique words: {unique_words}")
    print(f"Minimum frequency of a word: {min_freq}")
    print(f"Maximum frequency of a word: {max_freq}")
    print(f"Mean frequency: {mean_freq:.2f}")
    print(f"Median frequency: {median_freq}")
    print(f"Standard deviation of frequency: {std_freq:.2f}")

   # Histogram of word frequencies per page
    plt.figure(figsize=(8, 5))
    plt.hist(frequencies, bins=range(1, max_freq+2), edgecolor='black', alpha=0.7)
    plt.xlabel("Word frequency")
    plt.ylabel("Number of words")
    plt.title(f"Histogram of Word Frequencies - Page {page_no}")
    plt.tight_layout()
    plt.show()
```

```python
##displaying the descriptive statistics in one table of 83 pages

from collections import Counter
import numpy as np
import pandas as pd   # make sure pandas is installed

# pages is assumed to be a list of 83 lists,
# where each inner list contains the words of that page.
# Example: pages = [page1_words, page2_words, ..., page83_words]

rows = []

for page_no, page_words in enumerate(pages, start=1):
    counter = Counter(page_words)
    freqs = np.array(list(counter.values()), dtype=float)

    total_words   = freqs.sum()
    unique_words  = len(counter)
    mean_freq     = freqs.mean()
    median_freq   = np.median(freqs)
    std_freq      = freqs.std(ddof=0)   # population std; use ddof=1 for sample
    min_freq      = freqs.min()
    max_freq      = freqs.max()
    var_freq      = freqs.var(ddof=0)

    rows.append({
        "Page": page_no,
        "Total Words": int(total_words),
        "Unique Words": int(unique_words),
        "Mean Frequency": round(mean_freq, 6),
        "Median Frequency": float(median_freq),
        "Std Dev": round(std_freq, 6),
        "Min Frequency": int(min_freq),
        "Max Frequency": int(max_freq),
        "Variance": round(var_freq, 6)
    })

# Create table for all 83 pages
df_stats = pd.DataFrame(rows)

# Print as a nice table
print(df_stats.to_string(index=False))

# If you also want to save to CSV/Excel:
df_stats.to_csv("page_stats_83_pages.csv", index=False)
# df_stats.to_excel("page_stats_83_pages.xlsx", index=False)
```

```python
#TASK 3

from collections import Counter
import matplotlib.pyplot as plt

# REQUIRED DEFINITIONS
TOTAL_PAGES = len(pages)
stock_words = ["admission", "program", "student", "fee", "university"]

# ===============================
# PAGE-WISE WORD PROBABILITY
# ===============================
for page_no, page_words in enumerate(pages, start=1):
    counter = Counter(page_words)
    total_words = sum(counter.values())

    word_probabilities = {word: freq / total_words for word, freq in counter.items()}

    print("\n==============================")
    print(f"PAGE {page_no} WORD PROBABILITIES")
    print("==============================")

    for word, prob in word_probabilities.items():
        print(f"{word}: {prob:.4f}")

    print("\nStock Words Probability:")
    for sw in stock_words:
        prob = word_probabilities.get(sw, 0)
        print(f"{sw}: Frequency = {counter.get(sw, 0)}, Probability = {prob:.4f}")

# ===============================
# TASK 3: PLOTTING
# ===============================
stock_words_prob = {sw: [] for sw in stock_words}

for page_words in pages:
    counter = Counter(page_words)
    total_words = sum(counter.values())
    for sw in stock_words:
        stock_words_prob[sw].append(counter.get(sw, 0) / total_words)

plt.figure(figsize=(15, 6))
for sw, probs in stock_words_prob.items():
    plt.plot(range(1, TOTAL_PAGES + 1), probs, marker='o', label=sw)

plt.xlabel("Page Number")
plt.ylabel("Probability")
plt.title("Task 3: Stock Words Probability per Page")
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
#TASK3 
# =========================================================
#  TOP WORDS — HIGHEST PROBABILITY PAGE + GRAPHS
# =========================================================

import math
import pandas as pd
import matplotlib.pyplot as plt
from collections import Counter

FILE_NAME = "uhs_text_final.txt"
TOTAL_PAGES = 83
TOP_N_WORDS = 15

# Load text
with open(FILE_NAME, "r", encoding="utf-8") as f:
    all_words = f.read().lower().split()

# Split into pages
words_per_page = math.ceil(len(all_words) / TOTAL_PAGES)
pages = [
    all_words[i * words_per_page : (i + 1) * words_per_page]
    for i in range(TOTAL_PAGES)
]

page_counters = [Counter(page) for page in pages]
page_totals = [sum(counter.values()) for counter in page_counters]

# Overall most frequent words
overall_counter = Counter(all_words)
top_words = overall_counter.most_common(TOP_N_WORDS)

results = []

for word, total_freq in top_words:
    best_page = 0
    best_count = 0
    best_probability = 0

    for i, counter in enumerate(page_counters):
        count = counter.get(word, 0)
        total_words = page_totals[i]
        probability = count / total_words if total_words > 0 else 0

        if probability > best_probability:
            best_probability = probability
            best_page = i + 1
            best_count = count

    results.append([
        word,
        total_freq,
        best_page,
        best_count,
        best_probability
    ])

summary_df = pd.DataFrame(
    results,
    columns=[
        "Word",
        "Total Frequency (Overall)",
        "Page with Highest Probability",
        "Word Count on That Page",
        "Highest Probability"
    ]
)

print("\n===================================================")
print("MOST USED WORDS & THEIR HIGHEST PROBABILITY PAGES")
print("===================================================")
print(summary_df)

# -------- GRAPH 1: Highest Probability per Word --------
plt.figure(figsize=(9, 5))
plt.bar(summary_df["Word"], summary_df["Highest Probability"])
plt.xlabel("Word")
plt.ylabel("Highest Probability")
plt.title("Highest Page-wise Probability of Top Words")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()

# -------- GRAPH 2: Word Count on Peak Page --------
plt.figure(figsize=(9, 5))
plt.bar(summary_df["Word"], summary_df["Word Count on That Page"])
plt.xlabel("Word")
plt.ylabel("Word Count")
plt.title("Word Count on Page of Highest Probability")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```
